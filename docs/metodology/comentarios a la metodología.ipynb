## Comment 1

**Indicator construction and the role of time-varying risk**

"The integration of disequilibrium and risk follows the logic that a given level of imbalance poses greater danger when external volatility is high. We implement this through a normalized vulnerability metric with risk amplification:

$$
V_{t}=\frac{-\mu_{t}}{\sigma_{t}} \times\left(\frac{\sigma_{t}}{\sigma_{\mathrm{ref}}}\right)^{\lambda},
$$

where $\sigma_{\text {ref }}$ is a reference risk level chosen as the sample median, and $\lambda \geq 0$ is an amplification parameter. ... The parameter $\lambda$ controls the degree of amplification; we set $\lambda=1.0$ in the baseline specification, implying proportional scaling, though robustness checks explore $\lambda \in[0.5,1.5]$."

**Message:** Setting $\lambda = 1$ in equation&nbsp;(8.1) yields
$$
V_t \;=\;\frac{-\mu_t}{\sigma_t}\Bigl(\frac{\sigma_t}{\sigma_{\text{ref}}}\Bigr)^{1}
\;=\;\frac{-\mu_t}{\sigma_{\text{ref}}},
$$
so the time-varying risk measure $\sigma_t$ cancels out.  In that case the baseline indicator no longer depends on external volatility, contradicting (i) the surrounding explanation that “a given level of imbalance poses greater danger when external volatility is high,” and (ii) the episode analysis in Section 9.3, which attributes the 2008-09 spike in vulnerability to a surge in $\sigma_t$.  Please clarify whether a different $\lambda$ was used in the empirical work or whether the intended baseline specification should have $\lambda\neq1$.

## Comment 3

**Definition of shocks for the external risk measure**

"What remains is to combine these elements with information about the external environment's riskiness. This is where the time-varying structure of external shocks becomes crucial. The VECM's structural innovations—residuals from the estimated system—represent unexpected movements in the external variables after conditioning on past information. The variance-covariance matrix of these innovations, estimated in rolling windows, captures both the volatility of individual external shocks and their degree of synchronization."

**Message:** Statement that the “VECM’s structural innovations … represent unexpected movements in the external variables” is potentially misleading.  
The residual vector produced by the VECM contains the GDP innovation, while the external-risk block Σ_t used later is four-dimensional (ToT, US10Y, G7 IP, EM spread) to match the sensitivity vector g.  Because the text never explains how the two innovations associated with the stationary exogenous controls are obtained or how the GDP innovation is excluded when Σ_t is formed, a reader may struggle to see how the 4 × 4 covariance matrix is constructed.  
Adding one sentence that (i) clarifies that Σ_t is computed from a stacked vector that includes residuals (or demeaned first-differences) of the two exogenous variables and (ii) makes explicit that the GDP residual is dropped would remove the ambiguity.

## Comment 6

**Predictive Performance: Misuse of AUROC terminology**

"Converting the continuous vulnerability measure into a binary early warning signal by flagging quarters when $V_{t}$ exceeds the 90th percentile, we assess classification performance for predicting negative growth quarters. The indicator achieves a true positive rate (sensitivity) of $68 \%$ and a false positive rate of $22 \%$, corresponding to an area under the ROC curve (AUROC) of 0.73 . This compares favorably to alternative binary predictors: a flag for when terms of trade growth is below its 10th percentile achieves AUROC of 0.61, a flag for when US yields are above their 90th percentile achieves 0.59, and a flag for when spreads exceed their 90 th percentile achieves 0.64 ."

**Message:** I initially thought the paragraph was mixing up the AUROC concept with performance at a single threshold, because only one (TPR, FPR) pair is reported.  On re-reading, however, I realised that the authors are evaluating a *binary* score (the 90th-percentile flag).  For a binary predictor the ROC curve has just two line segments and the AUROC collapses to $\tfrac12(1+\text{TPR}-\text{FPR})$, which equals 0.73 given the reported TPR = 0.68 and FPR = 0.22.  The same logic justifies the AUROC figures quoted for the other binary flags.  

Because some readers may not be aware of this simplification, it could help to insert a short note explaining that AUROC for a binary score reduces to $\tfrac12(1+\text{TPR}-\text{FPR})$.  Otherwise the terminology and numbers appear consistent.
